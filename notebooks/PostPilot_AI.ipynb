# PostPilot AI

Run the reproducible pipeline after placing `social_media_engagement.csv` in `data/`. The model deliberately excludes post-publication outcomes from its features.

```python
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd().parent / 'src'))
from postpilot_pipeline import load_and_prepare, train, add_predictions, strategy_summary

input_path = Path('../data/social_media_engagement.csv')
output_path = Path('../output')
df = load_and_prepare(input_path)
model, metrics, importance = train(df)
predictions = add_predictions(df, model)
metrics
```

```python
display(df.groupby('Platform')['Engagement_Rate'].mean().sort_values(ascending=False))
display(df.groupby('Topic Category')['Engagement_Rate'].mean().sort_values(ascending=False))
display(importance.head(15))
strategy_summary(df)
```

```python
output_path.mkdir(exist_ok=True)
predictions.to_csv(output_path / 'postpilot_predictions.csv', index=False)
importance.to_csv(output_path / 'feature_importance.csv', index=False)
print('Exported Power BI input files to', output_path.resolve())
```
